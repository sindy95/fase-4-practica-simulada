```python
from abc import ABC, abstractmethod
import datetime

# ------------------ LOGS ------------------
def registrar_log(mensaje):
    with open("logs.txt", "a") as archivo:
        archivo.write(f"{datetime.datetime.now()} - {mensaje}\n")


# ------------------ EXCEPCIONES ------------------
class ErrorSistema(Exception):
    pass


class ClienteInvalido(ErrorSistema):
    pass


class ServicioNoDisponible(ErrorSistema):
    pass


class ReservaError(ErrorSistema):
    pass


# ------------------ CLASE ABSTRACTA ------------------
class Entidad(ABC):

    @abstractmethod
    def mostrar_info(self):
        pass


# ------------------ CLIENTE ------------------
class Cliente(Entidad):

    def __init__(self, nombre, edad):

        if not nombre:
            raise ClienteInvalido("Nombre vacío")

        if edad <= 0:
            raise ClienteInvalido("Edad inválida")

        self.__nombre = nombre
        self.__edad = edad

    def mostrar_info(self):
        return f"Cliente: {self.__nombre}, Edad: {self.__edad}"


# ------------------ SERVICIO ABSTRACTO ------------------
class Servicio(ABC):

    def __init__(self, nombre):
        self.nombre = nombre

    @abstractmethod
    def calcular_costo(self):
        pass


# ------------------ SERVICIOS ------------------
class ReservaSala(Servicio):

    def __init__(self, horas):
        super().__init__("Reserva de Sala")
        self.horas = horas

    def calcular_costo(self):
        return self.horas * 50000


class AlquilerEquipo(Servicio):

    def __init__(self, dias):
        super().__init__("Alquiler de Equipo")
        self.dias = dias

    def calcular_costo(self):
        return self.dias * 30000


class Asesoria(Servicio):

    def __init__(self, horas):
        super().__init__("Asesoría")
        self.horas = horas

    def calcular_costo(self):
        return self.horas * 80000


# ------------------ RESERVA ------------------
class Reserva:

    def __init__(self, cliente, servicio):

        if not isinstance(cliente, Cliente):
            raise ReservaError("Cliente inválido")

        if not isinstance(servicio, Servicio):
            raise ReservaError("Servicio inválido")

        self.cliente = cliente
        self.servicio = servicio
        self.estado = "Pendiente"

    def confirmar(self):  
                                                 # prueba de commit 1 Majurehy Lilian Coy Ruiz  "Aqui coloco una opcion de un cambio para implementar
                                                 # Try, except, else, finally, encadenamiento de excepciones
                                                 # y manejo robusto de errores.

        try:
            # calculando el costo del servicio prueba
            costo = self.servicio.calcular_costo()

        except Exception as e:
            # Este error se guarda en el archivo logs.txt
            registrar_log(f"error al confirmar reserva: {e}")

            # Encadenamiento de excepciones
            raise ReservaError(
                "No se pudo confirmar la reserva"
            ) from e

        else:
            self.estado = "Confirmada"

            print(f"Reserva hecha correctamente. Precio: {costo}")

            registrar_log("Reserva confirmada")

        finally:
            registrar_log("Proceso de confirmacion realizado")

    def cancelar(self):
        self.estado = "Cancelada"
        print("Reserva cancelada")


# ------------- SISTEMA DE RESERVAS --------------------
class SistemaReservas:
                                                                   # Prueba de commit 2 Majurehy Lilian Coy Ruiz sistema de reservas con listas sin utilizar bases de datos

    def __init__(self):

        # lista para guardar clientes
        self.clientes = []

        # lista para guardar reservas
        self.reservas = []

    # Ingresar el cliente al sistema
    def agregar_cliente(self, cliente):

        if not isinstance(cliente, Cliente):
            raise ClienteInvalido(
                "Solo se puede ingresar clientes"
            )

        self.clientes.append(cliente)

        registrar_log("Ingreso de cliente exitoso")

    # Reservas
    def agregar_reserva(self, reserva):

        if not isinstance(reserva, Reserva):
            raise ReservaError(
                "Solo reservas validas"
            )

        self.reservas.append(reserva)

        registrar_log("Reserva ingresada correctamente")

    # Mostrar clientes
    def mostrar_clientes(self):

        print("\nClientes registrados")

        for cliente in self.clientes:
            print(cliente.mostrar_info())

    # Reserva ya realizada
    def mostrar_reservas(self):

        print("\nReservas realizadas")

        for reserva in self.reservas:

            print(
                f"Cliente: {reserva.cliente.mostrar_info()} | "
                f"Servicio: {reserva.servicio.nombre} | "
                f"Estado: {reserva.estado}"
            )


# ------------------ SIMULACIÓN ------------------
def simulacion():
    # commit 2 Majurehy Lilian Coy Ruiz implementando listas como memoria
    # para no utilizar bases de datos

    # Crear el sistema principal
    sistema = SistemaReservas()

    # ---------------- RESERVA VÁLIDA ----------------
    try:

        c1 = Cliente("Sindy", 25)

        # Guardar cliente en la lista
        sistema.agregar_cliente(c1)

        s1 = ReservaSala(2)

        r1 = Reserva(c1, s1)

        # Guardar reserva en la lista
        sistema.agregar_reserva(r1)

        r1.confirmar()

    except Exception as e:

        registrar_log(str(e))

        print("Error:", e)

    # ---------------- ERROR INTENCIONAL ----------------
    try:

        c2 = Cliente("", -5)  # cliente inválido

        sistema.agregar_cliente(c2)

    except Exception as e:

        registrar_log(str(e))

        print("Error controlado:", e)

    # ---------------- OTRA RESERVA VÁLIDA ----------------
    try:

        s2 = AlquilerEquipo(3)

        r2 = Reserva(c1, s2)

        sistema.agregar_reserva(r2)

        r2.confirmar()

    except Exception as e:

        registrar_log(str(e))

        print("Error:", e)

    # ---------------- RESERVA INCORRECTA ----------------
    try:

        s3 = Asesoria(1)

        r3 = Reserva("No cliente", s3)  # error intencional

        sistema.agregar_reserva(r3)

    except Exception as e:

        registrar_log(str(e))

        print("Error:", e)

    # ---------------- Muestra datos de clientes y reservas ----------------
    sistema.mostrar_clientes()

    sistema.mostrar_reservas()


# ------------------ EJECUCIÓN ------------------
simulacion()
```
